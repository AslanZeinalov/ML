### 1.2. В чем разница между PCA и SVD?

**Определение**

- **PCA (Principal Component Analysis)**:
  - Метод анализа данных, который используется для уменьшения размерности.
  - Основная цель — найти новые переменные (главные компоненты), которые объясняют как можно больше вариации данных.
  - Главные вектора(компоненты) ортогональны друг другу и упорядочены по убыванию собственных значений.

- **SVD (Singular Value Decomposition)**:
  - Алгебраический метод разложения матрицы $A$ на три матрицы:
    A = U Σ Vᵀ
    где:
    - $U$: ортогональная матрица левых сингулярных векторов ($m \times m$).
    - $\Sigma$: диагональная матрица сингулярных значений ($m \times n$).
    - $V^T$: ортогональная матрица правых сингулярных векторов ($n \times n$).

---

**Цель и применение**

- **PCA**:
  - Уменьшение размерности данных при сохранении как можно большей вариации.
  - Используется для:
    - Визуализации данных,
    - Устранения мультиколлинеарности,
    - Уменьшения шума в данных.
  - PCA ищет линейные комбинации исходных признаков, которые лучше всего объясняют данные.

- **SVD**:
  - Универсальный метод разложения матриц, применяется для:
    - Рекомендательных систем,
    - Обработки сигналов и изображений,
    - Вычисления псевдообратной матрицы,
    - Приближения матриц и других задач линейной алгебры.

---

**Математическая связь**

- PCA тесно связана с SVD:
  - Сначала вычисляется ковариационная матрица:  
    $cov(x,y) = \frac{\sum\limits_{i=1}^{n}(x_{i}-\bar{x})(y_{i}-\bar{y})}{N}$  


    $
    A = \begin{bmatrix}
        C_{aa} & C_{ab} & C_{ac} \\
        C_{ab} & C_{bb} & C_{bc} \\
        C_{ac} & C_{bc} & C_{cc} \\
    \end{bmatrix}
    $
  - Затем находятся собственные значения и собственные векторы матрицы $C$.  
  
    $Ф_{A}(λ) = |A-λE| = 0$ (например: $-λ^{3}+λ^{2}+2λ=0$, (A - матрица ковариаций в нашем случае)   
    $AV_{i} = λ_{i}V{i}$ ($V_{i}$≠0), $V_{i}$ - собственный вектор,  
    
   - Затем строится матрица меньшего размера:  
    $Ф_{A}(λ) = |A-λE| = 0$ (например: $-λ^{3}+λ^{2}+2λ=0$, (A - матрица ковариаций в нашем случае)  
    $AV_{i} = λ_{i}V{i}$ ($V_{i}$≠0), $V_{i}$ - собственный вектор,    
    
  
  - SVD разлагает матрицу $A$ следующим образом:
    A = U Σ Vᵀ,
    где:
    - $def(A^{T}*A-λI) = 0$ => находим корень из собственного значения $\sqrt{λ_{1}} =σ_{1}$ => $def(A^{T}A-λI)v = 0$
    - $def(A*A^{T}-λI) = 0$ => находим корень из собственного значения $\sqrt{λ_{1}} =σ_{1}$ => $def(A*A^{T}-λI)u = 0$  
    - S — диагональной матрицы с сингулярными значениями $\Sigma$  
    
---

**Интерпретация результата**

- **PCA**:
  - Даёт главные компоненты (новые оси данных).
  - Оценивает, сколько информации сохраняет каждая главная компонента.

- **SVD**:
  - Выводит три матрицы:
    - $U$: содержит направления данных в пространстве объектов.
    - $\Sigma$: диагональная матрица, представляющая "вес" или значимость каждого направления.
    - $V^T$: преобразует исходные признаки в новое пространство.

---

## 5. Ключевые различия

| **Характеристика**        | **PCA**                                               | **SVD**                         |
|----------------------------|-------------------------------------------------------|----------------------------------|
| **Цель**                  | Уменьшение размерности, нахождение главных компонент. | Алгебраическое разложение матрицы. |
| **Выход**                 | Главные компоненты (новые признаки).                 | Матрицы $U$, $\Sigma$, $V^T$. |
| **Использует SVD?**       | Да, для вычислений.                                   | Нет, это само по себе разложение. |
| **Применение**            | Анализ данных, визуализация, уменьшение размерности. | Обработка матриц, аппроксимация, системы уравнений. |

---



- PCA — это более прикладной метод, который использует SVD для уменьшения размерности.
- SVD — это универсальный математический инструмент, из которого PCA является частным случаем.

### 1.3. В чем разница между NMF и SVD?

### 1. Определение

- **SVD (Разложение сингулярных чисел):**  
  SVD разлагает матрицу $A$ (размера $m \times n$) на три матрицы:
  $$
  A = U \Sigma V^T
  $$
  где:
  - $U$ — ортогональная матрица ($m \times m$),
  - $\Sigma$ — диагональная матрица сингулярных значений ($m \times n$),
  - $V^T$ — ортогональная матрица ($n \times n$).  

- **NMF (Факторизация неотрицательных матриц):**  
  NMF разлагает матрицу $A$ (размера $m \times n$) на две матрицы:
  $$
  A \approx WH
  $$
  где:
  - $W$ — матрица (размера $m \times k$), содержащая только неотрицательные элементы,
  - $H$ — матрица (размера $k \times n$), также с неотрицательными элементами.

---

### 2. Цели и особенности

| Характеристика              | **SVD**                                    | **NMF**                                  |
|-----------------------------|--------------------------------------------|------------------------------------------|
| **Цель**                   | Полное или приближенное разложение матрицы | Приближенное разложение для неотрицательных данных |
| **Тип элементов**          | Любые числа (положительные и отрицательные)| Только неотрицательные элементы          |
| **Интерпретация**           | Применимо к широкому кругу задач           | Часто используется для интерпретации "частей" данных |
| **Ортогональность**        | Результирующие матрицы ортогональны        | Результирующие матрицы не ортогональны   |

---

### 3. Алгоритм и ограничения

- **SVD:**
  - Раскладывает матрицу с использованием линейной алгебры.
  - Не накладывает ограничений на знаки элементов матрицы.

- **NMF:**
  - Основана на оптимизационных алгоритмах (например, мультипликативные обновления, метод градиентного спуска).
  - Накладывает сильное ограничение на неотрицательность элементов, что полезно для интерпретации.

---

### 4. Пример использования

### **SVD:**
Представим, что у нас есть матрица данных пользователей и оценок фильмов:

$$
A =
\begin{bmatrix}
5 & 3 & 0 \\
4 & 0 & 0 \\
1 & 1 & 0 \\
0 & 0 & 5 \\
0 & 0 & 4
\end{bmatrix}
$$

SVD разложит эту матрицу на:
$$
A = U \Sigma V^T
$$

- $U$ содержит информацию о пользователях.
- $\Sigma$ указывает на "вес" каждого латентного фактора.
- $V^T$ содержит информацию о фильмах.

---

### **NMF:**
Та же матрица $A$ может быть разложена на:
$$
A \approx WH
$$

- $W$ содержит информацию о пользователях и их склонностях к определённым "темам" (например, жанрам фильмов).
- $H$ показывает, насколько сильно фильмы связаны с этими "темами".

Пример:
$$
W =
\begin{bmatrix}
0.8 & 0.2 \\
0.7 & 0.1 \\
0.1 & 0.9 \\
0.0 & 0.8 \\
0.0 & 0.7
\end{bmatrix}, \quad
H =
\begin{bmatrix}
5 & 3 & 0 \\
0 & 0 & 5
\end{bmatrix}
$$

---

### 5. Применение

| Применение                   | **SVD**                                      | **NMF**                                |
|------------------------------|----------------------------------------------|----------------------------------------|
| **Рекомендательные системы** | Сжатие данных для создания латентных факторов | Анализ предпочтений с неотрицательными данными |
| **Снижение размерности**    | Используется для PCA                         | Снижает размерность с интерпретацией  |
| **Текстовая аналитика**     | Анализ латентных семантик (LSA)              | Тематическое моделирование (LDA, NMF) |
| **Обработка изображений**   | Сжатие и аппроксимация                       | Сегментация объектов                   |

---

### 6. Ключевые различия

1. **Ограничение на неотрицательность:**
   - SVD работает с любыми значениями.
   - NMF требует, чтобы все значения были неотрицательными.

2. **Интерпретация:**
   - NMF даёт разложения, которые легко интерпретировать как "части".
   - SVD более общая техника, но интерпретация может быть сложной.

3. **Ортогональность:**
   - В SVD матрицы $U$ и $V$ ортогональны.
   - В NMF матрицы $W$ и $H$ не ортогональны, что делает их более гибкими, но менее математически строгими.

---

NMF чаще используется, когда интерпретация данных имеет значение (например, в текстовом анализе или обработке изображений), тогда как SVD универсальна и применяется для анализа любых данных, включая отрицательные.

### 1.4. Описать устройство алгоритма уменьшения размерности Locally Linear Embedding

### Locally Linear Embedding (LLE)

Locally Linear Embedding (LLE) — это метод нелинейного уменьшения размерности, который сохраняет локальную геометрию данных в высокоразмерном пространстве и проецирует её в пространство меньшей размерности. Он предполагает, что данные лежат на многообразии низкой размерности.

---

## Основные идеи
1. **Локальная линейность:**
   - Точки, которые близки друг к другу в исходном пространстве, остаются близки друг к другу в низкоразмерном представлении.
   - Каждая точка представляется как линейная комбинация своих соседей.

2. **Глобальная структура:**
   - Сохраняется локальная линейная геометрия данных, которая используется для построения низкоразмерного представления.

---

## Этапы работы алгоритма

### Этап 1: Поиск ближайших соседей
Для каждой точки данных $x_i$ находим $k$ ближайших соседей с помощью метрик расстояния (например, евклидовой метрики). Множество ближайших соседей обозначается как $\mathcal{N}(x_i)$.

### Этап 2: Построение матрицы весов
Для каждой точки $x_i$ вычисляется весовая матрица $W$, которая отражает вклад каждого соседа $x_j$ в реконструкцию $x_i$.

Решается следующая задача минимизации:

$$
\text{min}_W \sum_{i=1}^N \|x_i - \sum_{j \in \mathcal{N}(x_i)} w_{ij} x_j \|^2,
$$

при условиях:
- $w_{ij} = 0$, если $j \notin \mathcal{N}(x_i)$,
- $\sum_{j \in \mathcal{N}(x_i)} w_{ij} = 1$.

Каждая точка $x_i$ аппроксимируется линейной комбинацией её соседей.

### Этап 3: Построение низкоразмерного пространства
Ищется представление $Y = \{y_1, y_2, \dots, y_N\}$, где каждая точка $y_i \in \mathbb{R}^d$ (меньшей размерности $d$), сохраняя локальные отношения, заданные весами $W$.

Решается следующая задача минимизации:

$$
\text{min}_Y \sum_{i=1}^N \|y_i - \sum_{j \in \mathcal{N}(x_i)} w_{ij} y_j \|^2,
$$

при условиях:
1. $Y$ центрировано (среднее значение координат равно 0):  
   $$
   \sum_{i=1}^N y_i = 0,
   $$
2. $Y$ имеет единичную ковариационную матрицу:  
   $$
   \frac{1}{N} \sum_{i=1}^N y_i y_i^T = I.
   $$

Решение задачи минимизации осуществляется через вычисление собственных векторов матрицы:

$$
M = (I - W)^T (I - W),
$$

где $I$ — единичная матрица. Собственные векторы, соответствующие $d$ наименьшим ненулевым собственным значениям, формируют представление $Y$.

---

## Пример (упрощённый)

### Дано:
Три точки в 2D-пространстве:  
$x_1 = (1, 1)$, $x_2 = (2, 1)$, $x_3 = (1, 2)$.

### Этап 1: Поиск соседей
При $k=2$ ближайших соседей для каждой точки:
- Для $x_1$: соседи — $x_2$, $x_3$.
- Для $x_2$: соседи — $x_1$, $x_3$.
- Для $x_3$: соседи — $x_1$, $x_2$.

### Этап 2: Построение весов
Вычисляем веса $w_{ij}$ для каждой точки:
- Для $x_1$:  
  $$
  x_1 \approx w_{12} x_2 + w_{13} x_3.
  $$
  Решаем задачу:  
  $$
  \|x_1 - (w_{12} x_2 + w_{13} x_3)\|^2, \quad w_{12} + w_{13} = 1.
  $$
  Предположим, решение: $w_{12} = 0.5$, $w_{13} = 0.5$.

Аналогично для $x_2$ и $x_3$.

### Этап 3: Построение низкоразмерного пространства
- Строим матрицу $W$, содержащую веса.
- Вычисляем матрицу $M = (I - W)^T (I - W)$.
- Находим собственные векторы $M$, соответствующие наименьшим собственным значениям (кроме нулевого).

---

## Итог
После выполнения алгоритма получается низкоразмерное представление $Y$, которое сохраняет локальные геометрические свойства исходных данных.

### 2.2. Создайте матрицу взаимодействий пользователя с книгой. Используйте разреженные матрицы

### LocallyLinearEmbedding

Суть: Нелинейный метод. Предполагает, что данные лежат на некотором многообразии. Он пытается восстановить каждую точку как линейную комбинацию её ближайших соседей.

### Как работает алгоритм Locally Linear Embedding (LLE)?

1. **Нахождение ближайших соседей**:
   
   Для каждой точки в данных (объекта) мы находим ее $ k $-ближайших соседей. Это делается с помощью стандартных методов, таких как вычисление евклидовых расстояний между точками.

2. **Построение линейных коэффициентов**:
   
   Каждую точку представляют как линейную комбинацию ее ближайших соседей. Это делается таким образом, чтобы минимизировать ошибку реконструкции:
   
   $$
   x_i \approx \sum_{j \in N(i)} w_{ij} x_j
   $$
   
   где $ x_i $ — это точка $ i $, а $ x_j $ — соседи, $ w_{ij} $ — коэффициенты линейной комбинации. Эти коэффициенты $ w_{ij} $ можно найти путем решения оптимизационной задачи для каждой точки, минимизируя разницу между точкой и линейной комбинацией ее соседей.

3. **Минимизация потерь при проекции**:

   После того как для всех точек вычислены линейные коэффициенты, алгоритм пытается найти такое низкоразмерное представление точек (например, в 2D или 3D), которое сохраняет эти локальные линейные связи. То есть, для каждой точки $ i $ в новом пространстве должно выполняться следующее условие:
   
   $$
   y_i \approx \sum_{j \in N(i)} w_{ij} y_j
   $$
   
   где $ y_i $ — это новая проекция точки $ x_i $ в низкоразмерное пространство, а $ y_j $ — это новые проекции соседей. Оптимизация заключается в минимизации этой ошибки для всех точек, что приводит к сохранению локальной структуры данных.

4. **Поиск проекции с помощью собственных значений**:

   Для выполнения оптимизации используется метод, аналогичный задаче собственного разложения (eigenproblem). Для этого строится матрица графа на основе линейных коэффициентов $ w_{ij} $, и затем решается собственная задача для этой матрицы.

```python
lle = LocallyLinearEmbedding(n_components=2, n_neighbors=10, tol=1e-06, neighbors_algorithm='kd_tree', method='modified', eigen_solver='dense')
X_lle = lle.fit_transform(X_sub)

plt.figure(figsize=(8, 6))
scatter = plt.scatter(X_lle[:, 0], X_lle[:, 1], c=y_sub, cmap="tab10", alpha=0.5)
plt.colorbar(scatter, label="Digit Label")
plt.xlabel("SVD_rand Component 1")
plt.ylabel("SVD_rand Component 2")
plt.title("MNIST в 2D после SVD_rand")
plt.show()
```

### 3.5. Предложить метрику и сравнить модели на основе того, насколько хорошо они разделяют классы цифр

Все метрики для DR делятся на два больших класса:
1. Внутренние (Internal): Оценивают только геометрию точек. Им не нужны метки классов (цифры). Они смотрят, насколько плотно и далеко друг от друга лежат точки.
2. Внешние (External): Используют известные нам метки классов. Они оценивают, насколько хорошо новая геометрия сохраняет разделение между известными классами.

**Silhouette Score (Индекс силуэта) — Внутренняя метрика**

**Интуиция:**
Метрика отвечает на вопрос: "Насколько объект похож на объекты своего кластера по сравнению с объектами других кластеров?"
Для каждой точки вычисляются два значения:

**a** — среднее расстояние от этой точки до всех других точек в том же кластере (насколько она близко к своим).

**b** — среднее расстояние от этой точки до всех точек ближайшего чужого кластера (насколько она далеко от чужих).

Формула для одной точки:

$$ S = \frac{b - a}{max(a, b)} $$

Итоговый Silhouette Score — это среднее
$S$ по всем точкам.

Шкала: от -1 до 1.

- Близко к 1: Отлично. Точка далеко от чужих и близко к своим.

- Близко к 0: Точка находится на границе двух кластеров.

- Близко к -1: Плохо. Точка попала в чужой кластер.

**Точность 1-NN (1-Nearest Neighbor Accuracy) — Внешняя метрика-прокси**

**Интуиция:**
Это не совсем метрика кластеризации, это прокси-метрика (замена) для оценки разделимости классов.
Мы берем наше 2D-пространство и задаем вопрос: "Если я возьму случайную точку, и посмотрю на её самого близкого соседа в этом 2D-пространстве, насколько вероятно, что у них одинаковая цифра?"
Почему это крутая метрика для снижения размерности:
Она напрямую измеряет то, что нам нужно: сохранность локальной структуры классов. Если в 2D-пространстве классы образуют четкие, непересекающиеся "островки", то 1-NN покажет 100% точность. Если классы накладываются друг на друга, точность упадет.

**Как это работает на практике:**

1. Берем 2D-координаты (например, после UMAP).
2. Делим их на Train и Test.
3. Обучаем KNeighborsClassifier(n_neighbors=1) на Train.
4. Предсказываем метки для Test и считаем Accuracy (долю верных предсказаний).

**Как интерпретировать на MNIST:**

Это самая честная метрика для сравнения.
PCA/SVD: ~80-85% (многие цифры путаются, например, 3 и 8, 4 и 9).
t-SNE/UMAP: ~95-98% (островки цифр четко разделены, 1-NN почти не ошибается).


```python
from sklearn.metrics import silhouette_score
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import train_test_split

# Для метрики 1-NN разделим подвыборку (для нелинейных методов) на train и test
X_train, X_test, y_train, y_test = train_test_split(X_sub, y_sub, test_size=0.3, random_state=21)

results = {}

# Функция для оценки
def evaluate_projection(name, X_2d_full, X_2d_train, X_2d_test, y_full, y_train, y_test):
    sil_score = silhouette_score(X_2d_full, y_full)

    knn = KNeighborsClassifier(n_neighbors=1)
    knn.fit(X_2d_train, y_train)
    knn_accuracy = knn.score(X_2d_test, y_test)

    results[name] = {
        'Silhouette Score': sil_score,
        '1-NN Accuracy': knn_accuracy
    }

X_sub_train, X_sub_test, y_sub_train, y_sub_test = train_test_split(X_sub, y_sub, test_size=0.3, random_state=21)

X_pca_sub = PCA(n_components=2, random_state=21).fit_transform(X_sub)
evaluate_projection('PCA', X_pca_sub, X_pca_sub[X_sub_train.index if hasattr(X_sub_train, 'index') else np.arange(len(X_sub_train))],
                    X_pca_sub[np.arange(len(X_sub_train), len(X_sub))], y_sub, y_sub_train, y_sub_test)

evaluate_projection('t-SNE', X_tsne, X_tsne, X_tsne, y_sub, y_sub, y_sub)
evaluate_projection('UMAP', X_umap, X_umap, X_umap, y_sub, y_sub, y_sub)
evaluate_projection('LLE', X_lle, X_lle, X_lle, y_sub, y_sub, y_sub)

df_results = pd.DataFrame(results).T
print("\nИтоговая таблица сравнения:")
print(df_results)
```

## 4. Сжатие изображений с помощью SVD

1. Выберите 3 любых изображения в оттенках серого и загрузите их с помощью Python в матрицу
2. Разложите матрицу с помощью SVD
3. Изменяйте ранг, вычисляйте низкоранговое приближение исходной матрицы и стройте восстановленное изображение.
Используйте до 10 различных значений ранга. Как выглядит спектр сингулярных значений
 и сколько мод необходимо для хорошей реконструкции изображения?
4. Постройте распределение объясненной дисперсии по рангу. Объясните, что такое “explained variance”

```python
import numpy as np
import matplotlib.pyplot as plt
from skimage import data


# ПУНКТ 1: Загрузка 3 изображений

# Используем встроенные черно-белые изображения из skimage
images = [
    ("Camera", data.camera()),
    ("Moon", data.moon()),
    ("Coins", data.coins())
]

# Задаем 10 различных рангов (k) для эксперимента
# Ранг — это количество сингулярных значений, которые мы оставляем
ranks_to_test = [2, 5, 10, 20, 40, 80, 120, 160, 200, 250]
g = 0

for name, img in images:

    print(f"Обрабатываем изображение: {name} ({img.shape})")


    # ПУНКТ 2: Сингулярное разложение (SVD)

    # full_matrices=False экономит память, возвращая только нужные части U и Vt
    U, S, Vt = np.linalg.svd(img, full_matrices=False)

    # Создаем фигуру для отрисовки результатов
    fig, axes = plt.subplots(2, 5, figsize=(15, 6))
    fig.suptitle(f'Низкоранговое приближение: {name}', fontsize=16)

    # ПУНКТ 3: Изменение ранга и восстановление

    for i, k in enumerate(ranks_to_test):
        # Убеждаемся, что k не превышает размерность матрицы
        k = min(k, len(S))

        # Берем только первые k компонент
        U_k = U[:, :k]
        S_k = np.diag(S[:k]) # Превращаем вектор обратно в диагональную матрицу
        Vt_k = Vt[:k, :]

        # Восстанавливаем изображение (перемножаем матрицы)
        img_compressed = np.dot(U_k, np.dot(S_k, Vt_k))

        # Отрисовка
        ax = axes[i // 5, i % 5]
        ax.imshow(img_compressed, cmap='gray')
        ax.set_title(f'Rank k={k}')
        ax.axis('off')

    plt.tight_layout()
    plt.show()


    # ПУНКТ 4: Спектр сингулярных значений

    plt.figure(figsize=(10, 4))

    # График 1: Обычный спектр (линейная шкала)
    plt.subplot(1, 2, 1)
    plt.plot(S, color='blue')
    plt.title('Спектр сингулярных значений (Линейный)')
    plt.xlabel('Индекс компоненты (i)')
    plt.ylabel(r'Значение $\sigma_i$')
    plt.grid(True)

    # График 2: Спектр в логарифмической шкале
    plt.subplot(1, 2, 2)
    plt.plot(np.log(S), color='red')
    plt.title('Спектр (Логарифмическая шкала)')
    plt.xlabel('Индекс компоненты (i)')
    plt.ylabel(r'log($\sigma_i$)')
    plt.grid(True)

    plt.tight_layout()
    plt.show()


    # ПУНКТ 5: График объясненной дисперсии

    # Считаем для первой картинки (Camera)
    U, S, Vt = np.linalg.svd(images[g][1])

    # 1. Считаем квадраты сингулярных значений
    singular_values_squared = S ** 2

    # 2. Общая сумма энергии (100% информации)
    total_variance = np.sum(singular_values_squared)

    # 3. Накопленная (кумулятивная) сумма энергии для каждого ранга k
    explained_variance = np.cumsum(singular_values_squared)

    # 4. Переводим в проценты
    explained_variance_ratio = (explained_variance / total_variance) * 100

    # Построение графика
    plt.figure(figsize=(8, 5))
    plt.plot(range(1, len(S) + 1), explained_variance_ratio, marker='.', color='green')
    plt.axhline(y=90, color='r', linestyle='--', label='90% информации')
    plt.axhline(y=95, color='b', linestyle='--', label='95% информации')

    plt.title('Кумулятивная объясненная дисперсия (Explained Variance)')
    plt.xlabel('Ранг (количество сингулярных значений k)')
    plt.ylabel('Доля сохраненной информации (%)')
    plt.legend()
    plt.grid(True)
    plt.show()
    print(f"Ранг для сохранения 90% информации: {np.argmax(explained_variance_ratio >= 90) + 1}")
    print(f"Ранг для сохранения 95% информации: {np.argmax(explained_variance_ratio >= 95) + 1}")
    g += 1
```

**Объясненная дисперсия** (в процентах) показывает, какую долю от исходной информации мы сохранили, отбросив "мусорные" сингулярные значения.

**Вывод** По графикам видно, что в районе 25-50 мод необходимо для относительно хорошей реконструкции изображения

## 5. Распознавание фона с помощью SVD

1. Загрузить «Video_008.avi» (источник - bmc_real.zip)
2. Представить видео в виде матрицы и разложить ее с помощью SVD
3. Восстановить первый кадр с помощью низкорангового приближения. Какой ранг следует использовать для получения фона?

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import urllib
import glob
import zipfile

video_url = "https://backgroundmodelschallenge.eu/data/real/Video_008.zip"
zip_filename = "Video_008.zip"
target_video_name = "Video_008.avi"

# Скачиваем архив
print(f"Загрузка: {video_url}")
urllib.request.urlretrieve(video_url, zip_filename)
print("Загрузка завершена. Распаковка...")

# Распаковываем архив в текущую директорию
with zipfile.ZipFile(zip_filename, 'r') as zip_ref:
    zip_ref.extractall('.')
print("Распаковка завершена успешно!")

# Находим точный путь к .avi файлу (на случай, если он попал в подпапку при распаковке)
avi_files = glob.glob(f"**/{target_video_name}", recursive=True)
video_path = avi_files[0] if avi_files else target_video_name
print(f"Используемый путь к видео: {video_path}\n")

cap = cv2.VideoCapture(video_path)

frames = []
max_frames = 100 # Ограничиваем количество кадров

print("Чтение кадров из видео...")
while len(frames) < max_frames:
    ret, frame = cap.read()
    if not ret:
        break

    # 1. Переводим в оттенки серого (упрощает задачу и экономит память в 3 раза)
    gray_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    # 2. (Опционально) Можно уменьшить размер кадра для ускорения, например до 160x120
    gray_frame = cv2.resize(gray_frame, (160, 120))

    # 3. Разворачиваем 2D-кадр в 1D-вектор и добавляем в список
    frames.append(gray_frame.flatten())

cap.release()

# Создаем матрицу данных X.
# Размерность: (число_пикселей_в_кадре) x (количество_кадров)
X = np.column_stack(frames)
print(f"Размер матрицы видео: {X.shape} (пиксели x кадры)")

height, width = gray_frame.shape # Запоминаем размеры для обратного преобразования


# ШАГ 2: Сингулярное разложение (SVD)

print("Выполнение SVD...")

U, S, Vt = np.linalg.svd(X, full_matrices=False)


# ШАГ 3 и 4: Восстановление фона при разных рангах

ranks_to_test = [1, 3, 5, 10, 20, 50]

plt.figure(figsize=(15, 10))
plt.suptitle('Восстановление фона первого кадра при разных рангах (k)', fontsize=16)

for i, k in enumerate(ranks_to_test):

    # Берем первые k компонент
    U_k = U[:, :k]
    S_k = S[:k]
    Vt_k_0 = Vt[:k, 0] # 0-й столбец Vt соответствует первому кадру

    # Восстанавливаем вектор первого кадра: U_k * (S_k * Vt_k_0)

    frame_1_reconstructed = np.dot(U_k, S_k * Vt_k_0)

    # Преобразуем вектор обратно в 2D-изображение
    img_bg = frame_1_reconstructed.reshape(height, width)

    # Ограничиваем значения диапазоном [0, 255] и приводим к uint8 для отображения
    img_bg = np.clip(img_bg, 0, 255).astype(np.uint8)

    # Отрисовка
    plt.subplot(2, 3, i + 1)
    plt.imshow(img_bg, cmap='gray')
    plt.title(f'Ранг k = {k}')
    plt.axis('off')

plt.tight_layout()
plt.show()

# ШАГ 5: Анализ спектра для выбора ранга

plt.figure(figsize=(8, 4))
plt.plot(S[:50], marker='o', color='purple') # Смотрим первые 50 значений
plt.title('Спектр сингулярных значений видео-матрицы')
plt.xlabel('Индекс сингулярного числа (k)')
plt.ylabel('Значение')
plt.grid(True)
plt.show()
```
