## 1. Аналитическое решение задач регресии

### 1.1 Общая формула

Аналитическое решение задачи регрессии можно представить в векторной форме уравнения:

$$
\mathbf{X}^\top \mathbf{X} \boldsymbol{\beta} = \mathbf{X}^\top \mathbf{y}
\quad \Rightarrow \quad
\boldsymbol{\beta} = (\mathbf{X}^\top \mathbf{X})^{-1} \mathbf{X}^\top \mathbf{y}
$$

**Обозначения символов:**
- $\mathbf{X} \in \mathbb{R}^{n \times m}$ — матрица признаков  
  ($n$ — число объектов, $m$ — число признаков; **первый столбец = 1** для свободного члена $\beta_0$)
- $\mathbf{y} \in \mathbb{R}^{n}$ — вектор целевой переменной (наблюдаемые значения)
- $\boldsymbol{\beta} \in \mathbb{R}^{m}$ — вектор коэффициентов модели:  
  $\boldsymbol{\beta} = \begin{bmatrix} \beta_0 & \beta_1 & \cdots & \beta_{m-1} \end{bmatrix}^\top$
- $\mathbf{X}^\top$ — транспонированная матрица $\mathbf{X}$
- $(\mathbf{X}^\top \mathbf{X})^{-1}$ — обратная матрица (существует при полном ранге $\mathbf{X}$)
- $\mathbb{R}^{a \times b}$ — пространство матриц размера $a \times b$ с вещественными элементами

**Функция потерь MSE (Mean Squared Error)**

$$
L(\boldsymbol{\beta}) = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2 = \frac{1}{n} \sum_{i=1}^{n} (y_i - \mathbf{x}_i^\top \boldsymbol{\beta})^2
$$

**В векторной форме:**

$$
L(\boldsymbol{\beta}) = \frac{1}{n} (\mathbf{y} - \mathbf{X}\boldsymbol{\beta})^\top (\mathbf{y} - \mathbf{X}\boldsymbol{\beta})
$$

**Обозначения:**
- $n$ — количество объектов в выборке
- $y_i$ — истинное значение целевой переменной
- $\hat{y}_i$ — предсказанное значение
- $\mathbf{x}_i$ — вектор признаков $i$-го объекта
- $\boldsymbol{\beta}$ — вектор коэффициентов модели
- $\mathbf{X}$ — матрица признаков
- $\mathbf{y}$ — вектор целевых значений

### 1.2 Регуляризация. L1(Lasso) и L2(Ridge)

При добавлении регуляризации(**штрафной член**) L1 и L2 в функцию потерь мы минимизируем не только ошибку предсказания, но и штрафуем "сложность модели", т.е. слишком большие веса у признаков. Это помогает избежать переобучения модели.

В **первом случае** модель отсеивает признаки, которые нерелевантны или избыточны, оставляя только те, которые важны для предсказания целевой переменной. **L1** не имеет аналитического решения, т.к. недифференцируема в точке w = 0, что делает невозможным получение замкнутой формулы через классическое приравнивание градиента к нулю.
    

**L1** используют, когда нужно **уменьшить колличество признаков** потому, что данная регуляризация обнуляет слабые признаки, а не просто уменьшает их. Это происходит из-за того, что L1 "штарфует" веса фиксированной ставкой.

**Формула функции потерь для L1:**

$$
L_{\text{Lasso}} = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2 + \alpha \sum_{j=1}^{m} |w_j|
$$

**Градиент:**

$$
\frac{\partial L}{\partial w_j} = \frac{2}{n} \sum_{i=1}^{n} (\hat{y}_i - y_i) \cdot x_{ij} + \alpha \cdot \text{sign}(w_j)
$$

где:
- $n$ — количество объектов в выборке
- $m$ — количество признаков
- $y_i$ — истинное значение для $i$-го объекта
- $\hat{y}_i$ — предсказанное значение для $i$-го объекта
- $w_j$ — вес $j$-го признака
- $\alpha$ — параметр регуляризации
- $x_{ij}$ — значение $j$-го признака для $i$-го объекта
- $\text{sign}(w_j)$ — функция знака ($+1$, $-1$ или $0$)

**Особенности:**
- Обнуляет неважные веса → отбор признаков
- Недифференцируема в точке $w_j = 0$
- Не имеет аналитического решения

Во **втором случае** модель "штрафует" веса пропорционально квадрату коэффициента, т.е. даже признак имеющий очень слабую корреляцию с целевой переменной остается, но имеет очень маленький коэффициент веса.

Аналитическое решение с добавленим регуляризации **L2**(Ridge/Гребневая регрессия):

$$
\boldsymbol{\beta} = (X^T X + \lambda I)^{-1} X^T y
$$

| Символ | Описание |
|--------|----------|
| $\boldsymbol{\beta}$ | Вектор оптимальных параметров модели (веса + смещение) |
| $X$ | Матрица признаков размера $(n_{\text{samples}}, n_{\text{features}})$<br>*(обычно с добавленным столбцом единиц для смещения)* |
| $y$ | Вектор целевых значений размера $(n_{\text{samples}},)$ |
| $\lambda$ | Параметр регуляризации (λ ≥ 0)<br>• λ = 0 → обычная линейная регрессия<br>• λ → ∞ → все веса стремятся к нулю |
| $I$ | Единичная матрица размера $(n_{\text{features}}, n_{\text{features}})$ |
| $X^T$ | Транспонированная матрица признаков |
| $(\cdot)^{-1}$ | Обратная матрица |

Формула штрафа:
$$
R(\boldsymbol{\beta}) = \sum_{j=1}^{m} \beta_j^2
$$
Формула потерь для L2:
$$
L_{\text{Ridge}} = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2 + \lambda \sum_{j=1}^{m} \beta_j^2
$$

### 1.3 Нелинейные зависимости

Эти же модели могут выявлять зависимости не только когда данные похожи на прямую. Для этого используются криволинейные функции - многочлены, а саму модель называют **полиноминальной регрессией**.

Процесс обучения модели полиноминальной регрессии аналогичен процессу обучения модели линейной регресии. Единственной отличие заключается в том, что нужно добавить больше столбцов в набор данных, прежде чем применять линейную регрессию, а именно значения признаков возведенные в необходимую степень.

Поделим данные на тренировочные и валидационные

```python
train_df, valid_df = train_test_split(
    train_df,
    test_size=0.2,
    random_state=21
)
```

## 4. Реализация моделей - Линейная регрессия

Реализуем класс для аналитического решения линейной регрессии:

```python
class Linreg:
    def __init__(self):
        self.weights = None

    def fit(self, X, y):
        X = np.c_[np.ones((X.shape[0], 1)), X]
        self.weights = np.linalg.pinv(X).dot(y)
        return self

    def predict(self, X):
        X = np.c_[np.ones((X.shape[0], 1)), X]
        return X.dot(self.weights)
```

Создаздим еще один вариант с добавлением стохастического градиентного спуска (SDG) для поиска оптимальных весовых коэффициентов модели:

```python
class LinregSDG:
    def __init__(self, learning_rate=0.1, n_epochs=150, batch_size=128, random_state=None):
        self.learning_rate = learning_rate
        self.n_epochs = n_epochs
        self.batch_size = batch_size
        self.random_state = random_state
        self.weights = None

    def fit(self, X, y):
        if self.random_state is not None:
            np.random.seed(self.random_state)

        X = np.asarray(X)
        y = np.asarray(y)

        X = np.c_[np.ones((X.shape[0], 1)), X]
        n_samples, n_features = X.shape

        self.weights = np.random.randn(n_features)

        for epoch in range(self.n_epochs):
            indices = np.random.permutation(n_samples)

            for i in range(0, n_samples, self.batch_size):

                batch_idx = indices[i:i + self.batch_size]
                X_batch = X[batch_idx]
                y_batch = y[batch_idx]

                predictions = X_batch.dot(self.weights)
                errors = predictions - y_batch
                gradient = (2 / len(y_batch)) * X_batch.T.dot(errors)
                self.weights -= self.learning_rate * gradient

        return self
    
    def predict(self, X):

        X = np.c_[np.ones((X.shape[0], 1)), X]
        return X.dot(self.weights)

    def r2_score(self, X, y):
        y_pred = self.predict(X)
        y_mean = np.mean(y)
        ss_res = np.sum((y - y_pred) ** 2)
        ss_tot = np.sum((y - y_mean) ** 2)
        return 1 - (ss_res / ss_tot)
```

Чтобы сделать модель детерминированной, нужно установить значение параметра random_state(seed) для фиксации случайности.

## 5. Реализация регуляризованных моделей - Ridge, Lasso, ElasticNet

```python
class BaseRegularizedLinreg:
    def __init__(self, alpha=0.01, l1_ratio=0.5, learning_rate=0.1, n_epoch=150, batch_size=64, random_state=None):
        self.alpha = alpha
        self.l1_ratio = l1_ratio
        self.learning_rate = learning_rate
        self.n_epoch = n_epoch
        self.batch_size = batch_size
        self.random_state = random_state
        self.theta = None

    def __add_bias(self, X):
    
        return np.c_[np.ones((X.shape[0], 1)), X]
    

    def _get_regularization_loss(self):
    
        raise NotImplementedError
    

    def _get_regularization_gradient(self, m):
    
        raise NotImplementedError
    

    def fit(self, X, y):

        if self.random_state is not None:
            np.random.seed(self.random_state)

        X = np.asarray(X, dtype=np.float64)
        y = np.asarray(y, dtype=np.float64)

        X_b = self.__add_bias(X)
        n_samples, n_features = X_b.shape
        self.theta = np.random.randn(n_features)

        for epoch in range(self.n_epoch):
            indices = np.random.permutation(n_samples)
            X_b_shuffled = X_b[indices]
            y_shuffled = y[indices]

            for i in range(0, n_samples, self.batch_size):
                X_batch = X_b_shuffled[i:i + self.batch_size]
                y_batch = y_shuffled[i:i + self.batch_size]

                m = len(y_batch)
                predictions = X_batch.dot(self.theta)
                erros = predictions - y_batch
                gradient = (2 / m) * X_batch.T.dot(erros)

                reg_gradient = self._get_regularization_gradient(m)
                gradient[1:] += reg_gradient

                self.theta -= self.learning_rate * gradient

        return self
    

    def predict(self, X):
        X_b = self.__add_bias(X)
        return X_b.dot(self.theta)
    

    def score(self, X, y):
        r2 = r2_score(y, self.predict(X))
        mse = mean_absolute_error(y, self.predict(X))
        rmse = np.sqrt(mean_squared_error(y, self.predict(X)))

        return r2, mse, rmse

    
    def get_params(self):

        return {
            'intercept': self.theta[0],
            'coefficients': self.theta[1:],
            'alpha': self.alpha
        }
```

```python
class LassoLinreg(BaseRegularizedLinreg):

    def _get_regularization_gradient(self, m):
        return (self.alpha / m) * np.sign(self.theta[1:])
```

```python
class RidgeLinreg(BaseRegularizedLinreg):

    def _get_regularization_gradient(self, m):
        return (2 * self.alpha / m) * self.theta[1:]
```

```python
class ElasticNetLinreg(BaseRegularizedLinreg):

    def __init__(self, alpha=0.1, l1_ratio=0.5, **kwargs):
        super().__init__(alpha=alpha, **kwargs)
        self.l1_ratio = l1_ratio

    def _get_regularization_gradient(self, m):
        l2_gradient = (2 * self.alpha * (1 - self.l1_ratio) / m) * self.theta[1:]
        l1_gradient = (self.alpha * self.l1_ratio / m) * np.sign(self.theta[1:])
        return l2_gradient + l1_gradient
    
    def get_params(self):
        params =  super().get_params()
        params['l1_ratio'] = self.l1_ratio
        return params
```

Сделаем прогнозы с помощью наших моделей и моделей из sklearn'a:

```python
models = {
    'my_Ridge': RidgeLinreg(alpha=1, learning_rate=0.1, n_epoch=150, random_state=21),
    'my_Lasso': LassoLinreg(alpha=1, learning_rate=0.1, n_epoch=150, random_state=21),
    'my_ElasticNet': ElasticNetLinreg(alpha=1, l1_ratio=0.5, learning_rate=0.1, n_epoch=150, random_state=21),
    'Ridge': Ridge(alpha=1, random_state=21, max_iter=1000),
    'Lasso': Lasso(alpha=1, random_state=21, max_iter=1000),
    'ElasticNet': ElasticNet(alpha=1, l1_ratio=0.5, random_state=21, max_iter=1000)
}
```

```python
def count_metrics(models, X_train, X_val, X_test, y_train, y_val, y_test, mae_df, rmse_df, r2_df):
    for name, model in models.items():
        model.fit(X_train, y_train)
        y_pred_train = model.predict(X_val)
        y_pred_test = model.predict(X_test)

        mae = [name, mean_absolute_error(y_val, y_pred_train), mean_absolute_error(y_test, y_pred_test)]
        rmse =  [name, np.sqrt(mean_squared_error(y_val, y_pred_train)), np.sqrt(mean_squared_error(y_test, y_pred_test))]
        r2 = [name, r2_score(y_val, y_pred_train), r2_score(y_test, y_pred_test)]

        mae_df = pd.concat([mae_df, pd.DataFrame([mae], columns=mae_df.columns)], ignore_index=True)
        rmse_df = pd.concat([rmse_df, pd.DataFrame([rmse], columns=rmse_df.columns)], ignore_index=True)
        r2_df = pd.concat([r2_df, pd.DataFrame([r2], columns=r2_df.columns)], ignore_index=True)

    return mae_df, rmse_df, r2_df
```

```python
mae_df, rmse_df, r2_df = count_metrics(
    models,
    X_train, X_val, X_test,
    y_train, y_val, y_test,
    mae_df, rmse_df, r2_df
    )
```

## 6. Нормализация функций

**Нормализация** - это процесс приведения признаков к единому маштабу. Без нее признаки с большими значениями могут доменировать над признаками с малыми значениями, что искажает работу модели.

Нормализация **обязательна** при работе с ***градиентным спуском***(без нормализации он будет "скакать"  по осям, обучение замедлится или не сойдется), ***регуляризацией***(если признаки будут в разных маштабах, то штраф будет несправедливым), методами основанными на расстояниях и др.

Нормализация **не нужна** в ситуациях, когда мы используем ***древовидные модели***(деревья разбивают по порогам, маштаб не влияет на разбиение), признаки уже в ***одном маштабе***, важна ***интерпретируемость***(нужно сохранить исходные единицы измерения)

### 6.1 MinMaxScaler

**MinMaxScaler** - метод нормализации, который преобразует признаки к диапазону [0,1] (или другому заданному диапазону)


$$ X_{scaled} = \frac{X - X_{min}}{X_{max} - X_{min}} $$

Где:
- **$X$** — исходное значение признака
- **$X_{min}$** — минимальное значение признака в обучающей выборке
- **$X_{max}$** — максимальное значение признака в обучающей выборке
- **$X_{scaled}$** — нормализованное значение (обычно в диапазоне [0, 1])

Если нужно вернуть данные к исходному масштабу:

$$ X_{original} = X_{scaled} \cdot (X_{max} - X_{min}) + X_{min} $$


Создадим собственную функцию для нормализации признаков с помощью MinMaxScaler

```python
class MyMinMaxScaler:
    def __init__(self, feature_range=(0, 1)):
        self.feature_range = feature_range
        self.min_ = None
        self.max_ = None
        self.scale_ = None
        self.n_feaatures_ = None


    def fit(self, X):

        X = np.asarray(X, dtype=np.float64)

        if X.ndim == 1:
            X = X.reshape(-1, 1)     

        self.n_feaatures_ = X.shape[1]

        self.min_ = np.min(X, axis=0)
        self.max_ = np.max(X, axis=0)

        data_range = self.max_ - self.min_

        data_range[data_range == 0] = 1.0

        self.scale_ =  (self.feature_range[1] - self.feature_range[0]) / data_range

        return self

    def transform(self, X):
        if self.min_ is None or self.max_ is None:
            raise ValueError("Вызовите fit() перед transform()")
        
        X = np.asarray(X, dtype=np.float64)

        if X.ndim == 1:
            X = X.reshape(-1, 1)

        if X.shape[1] != self.n_feaatures_:
            raise ValueError(f"Ожидается {self.n_feaatures_} признаков. Получено {X.shape[1]}")
        
        X_scaled = (X - self.min_) * self.scale_ + self.feature_range[0]

        return X_scaled
    
    def fit_transform(self, X):

        return self.fit(X).transform(X)
    
    def inverse_transform(self, X):
        if self.min_ is None or self.max_ is None:
            raise ValueError("Сначала вызовите fit()")
        
        X = np.asarray(X, dtype=np.float64)

        if X.ndim == 1:
            X = X.reshape(-1, 1)

        X_original = (X - self.feature_range[0]) / self.scale_ + self.min_

        return X_original
```

### 6.2 StandartScaler

**StandardScaler** — это метод нормализации, который преобразует признаки к распределению с средним 0 и стандартным отклонением 1.

$$ X_{scaled} = \frac{X - \mu}{\sigma} $$

Где:
- **$X$** — исходное значение признака
- **$\mu$** (mu) — среднее значение признака в обучающей выборке (mean)
- **$\sigma$** (sigma) — стандартное отклонение признака в обучающей выборке (std)
- **$X_{scaled}$** — нормализованное значение (среднее = 0, стд. отклонение = 1)





Если нужно вернуть данные к исходному масштабу:

$$ X_{original} = X_{scaled} \cdot \sigma + \mu $$

```python
class MyStandartScaler:
    def __init__(self):
        self.mean_ = None
        self.std_ = None
        self.n_features_ = None

    def fit(self, X):
        X = np.asarray(X, dtype=np.float64)

        if X.ndim ==1:
            X = X.reshape(-1, 1)

        self.n_features_ = X.shape[1]
        self.mean_ = np.mean(X, axis=0)
        self.std_ = np.std(X, axis=0)

        self.std_[self.std_ == 0] = 1

        return self
    
    def transform(self, X):
        if self.mean_ is None or self.std_ is None:
            raise ValueError("Сначала вызовите fit()")
        
        X = np.asarray(X, dtype=np.float64)

        if X.ndim == 1:
            X = X.reshape(-1, 1)

        X_scaled = (X - self.mean_) / self.std_

        return X_scaled
    
    def fit_transform(self, X):
        return self.fit(X).transform(X)
    
    def inverse_transform(self, X):
        if self.mean_ is None or self.std_ is None:
            raise ValueError("Сначала fit()")
        
        X = np.asarray(X, dtype=np.float64)

        if X.ndim == 1:
            X = X.reshape(-1, 1)

        X_original = X * self.std_ + self.mean_

        return X_original
```
